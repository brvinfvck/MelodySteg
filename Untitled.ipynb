```python
import librosa
from scipy.io import wavfile
import numpy as np
import matplotlib.pyplot as plt
```

```python
FREQS = np.array([220.00, 261.63, 293.66, 329.63,
                 392.00, 440.00, 523.25, 587.33])
```

```python
sr, audio = wavfile.read('msteg/orig/mensaje.wav')
sr, audio.size
```

```python
audio = audio.astype(np.float32)
# Mono
if len(audio.shape) == 2:
    audio = audio[:, 0]
audio
```

```python
y = librosa.util.normalize(audio)
```

```python
onsets = librosa.onset.onset_detect(y=y, sr=sr, units='samples', backtrack=False)
onsets
```

```python
plt.plot(y)
for onset in onsets:
    plt.axvline(onset, color='red')
```

```python
muestra = 0.25

frecs_encontradas = []
for onset in onsets:  # por cada nota, extrae la frec
    inicio = int(onset)
    fin = int(onset + muestra * sr)
    segmento = y[inicio:fin]
    fft = np.abs(np.fft.rfft(segmento))
    f = np.argmax(fft) * sr / segmento.size
    frecs_encontradas.append(f)

frecs_encontradas
```

```python
# test = bpf(mensaje)
fft = np.fft.rfft(audio)

min_freq = np.min(FREQS)
max_freq = np.max(FREQS)*8

# freq 2 fft.sample
# Si f = np.argmax(fft) * sr / segmento.size
min_samp = int(min_freq * audio.size / sr)
max_samp = int(max_freq * audio.size / sr)

filtered = fft.copy()
filtered[:min_samp] = 0
filtered[max_samp:] = 0

plt.plot(np.abs(fft))
plt.plot(np.abs(filtered))
test = np.fft.irfft(filtered).real.astype(np.float32)
```
