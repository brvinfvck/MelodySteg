# Peaks detection

```python
from scipy.signal import find_peaks

def find_peaks(s, n=1):
    
    # Next < Curr and Last < Curr 
    peaks = np.where(
        (s[1:] > s[:-1])[:-1]
        &
        (s[1:-1]>s[2:])
    )[0]
    
    # +1 (compared arrays are shifted 1)
    peaks += 1
    
    if n == 1:
        return peaks

    return peaks[find_peaks(s[peaks], n=n-1)]
```

```python
import matplotlib.pyplot as plt

t_peaks = find_peaks(l, n=1)

plt.plot(l[:500])
plt.plot(t_peaks[:75], l[t_peaks][:75], "x")
```

# Find peaks rate

~ onsets

```python
rates = []
def get_rate(l, n=1):
    peaks = find_peaks(l, n=n)
    # Distance between peaks
    rate = peaks[1:] - peaks[:-1]
    return rate

rates = []
for i in range(1, 10):
    rate = get_rate(l, n=i)/samp_rate
    rates.append(np.abs(rate.mean() - 0.25))
    print(i, rate.mean(), np.abs(rate.mean() - 0.25))
    
best_rate_n = np.argmin(rates) + 1
print(best_rate_n)
```

```python
window_size = int(rates[best_rate_n]*samp_rate)
bw = window_size//2

fft = np.fft.fft(l[window_size:2*window_size])

bw = fft.size//2
freqs = np.roll(np.arange(-bw, bw+(1 if bw%2 else 0)), bw) * (samp_rate/(2*bw))

max_freq = 10e6
max_i = int(max_freq / samp_rate)

plt.plot(freqs[:max_i], np.abs(fft)[:max_i])
```
