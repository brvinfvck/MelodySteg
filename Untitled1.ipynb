```python
num_frames = 1 + (len(sound) - frame_size) // hop_size
# hdf
res = np.zeros(num_frames)

# 2. Process Frames
for k in range(num_frames):

    start = k * hop_size
    frame = h * sound[start : start + frame_size]

    # FFT (Real FFT for efficiency)
    spec = np.abs(np.fft.rfft(frame))

    # 3. Calculate High Frequency Content
    # We multiply the magnitude by the bin index (weights higher freqs)
    bin_indices = np.arange(len(spec))
    res[k] = np.sum(spec * bin_indices)
    
plt.plot(res)
```

```python
odf = 1 - (res[:-1] - res[1:])
plt.plot(odf)
```

```python
odf = np.diff(res, prepend=res[0])
plt.plot(odf)
```

```python
odf = np.maximum(0, odf) # Half-wave rectification
plt.plot(odf)
```

```python
# 5. Peak Picking (Thresholding)
# Simple dynamic threshold: mean + 2x standard deviation
threshold = np.mean(odf) + (2 * np.std(odf))
threshold
```

```python
onsets_in_frames = np.where(odf > threshold)[0]
plt.plot(onsets_in_frames, '.')
```

```python
from scipy.stats import zscore
onsets_in_frames_alt = np.where(zscore(odf) > 1)[0]

plt.plot(onsets_in_frames, onsets_in_frames, '.', color='blue')
plt.plot(onsets_in_frames_alt, onsets_in_frames_alt, 'x', color='red')
```
